```python
import yaml
import pandas as pd

print("--- Starting Data Validation Test ---")
yaml_fp = "/home/htet-aung-lynn/Study/E2E-Network-Security-ETL-Pipeline/data_schema/schema.yaml"
train_fp = "/home/htet-aung-lynn/Study/E2E-Network-Security-ETL-Pipeline/artifacts/2026_08_04_23_21_23/data_ingestion/ingested/train.csv"
test_fp = "/home/htet-aung-lynn/Study/E2E-Network-Security-ETL-Pipeline/artifacts/2026_08_04_23_21_23/data_ingestion/ingested/test.csv"

def validate_cols(yaml_fp, file_path):
    # Load schema and data in one go
    with open(yaml_fp, 'r') as f:
        schema = yaml.safe_load(f).get("columns_dtypes", [])
    
    df = pd.read_csv(file_path)
    
    # Flatten schema list of dicts into a single dictionary: {'col_name': 'dtype'}
    # schema_dict = {col: dtype for item in schema for col, dtype in item.items()}
    schema_dct = {col: dtype for item in schema for col, dtype in item.items()}
    
    # Check for missing columns and type mismatches using compact generator expressions
    missed_col = [col for col in schema_dct if col not in df.columns]

    mistmatch_dtype = [col for col, exp_type in schema_dct.items() if exp_type != str(df[col].dtype)]

    # missing = [col for col in schema_dict if col not in df.columns]
    # mismatches = [col for col, exp_type in schema_dict.items() if col in df.columns and str(df[col].dtype) != exp_type]
    
    print(f"File: {file_path.split('/')[-1]} | Missing: {len(missed_col)} | Type Mismatches: {len(mistmatch_dtype)}")

# Test both files instantly
validate_cols(yaml_fp, train_fp)
# validate_cols(yaml_fp, test_fp)
```
